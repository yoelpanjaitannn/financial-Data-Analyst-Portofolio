````python
import pandas as pd
import numpy as np

data = {
    'Item': [
        'Pendapatan Bersih',
        'Beban Pokok Pendapatan',
        'Laba Kotor',
        'Beban Operasional',
        'Laba Sebelum Pajak',
        'Laba Bersih',
        'Total Aset',
        'Total Aset Lancar',
        'Total Aset Tidak Lancar',
        'Total Liabilitas',
        'Total Liabilitas Jangka Pendek',
        'Total Ekuitas',
        'Kas dan Setara Kas',
        'Piutang Usaha',
        'Persediaan'
    ],
    '2022': [
        20988297, 19258198, 1730099, 605269,
        1138595, 866721, 8582896, 7898375,
        684521, 4166033, 4028949, 4416863,
        1383372, 2407610, 1440347
    ],
    '2023': [
        22086390, 20175440, 1910950, 714484,
        1217695, 940104, 10146570, 9313853,
        832717, 5108434, 4849005, 5038136,
        1383556, 3292249, 1498768
    ]
}

df = pd.DataFrame(data).set_index('Item')
print("=== DATA LAPORAN KEUANGAN MTDL (Juta Rupiah) ===")
print(df.to_string())
````

````python
rasio = {}

for year in ['2022', '2023']:
    rasio[year] = {
        'Gross Profit Margin (%)': round(df.loc['Laba Kotor', year] / df.loc['Pendapatan Bersih', year] * 100, 2),
        'Net Profit Margin (%)': round(df.loc['Laba Bersih', year] / df.loc['Pendapatan Bersih', year] * 100, 2),
        'ROA (%)': round(df.loc['Laba Bersih', year] / df.loc['Total Aset', year] * 100, 2),
        'ROE (%)': round(df.loc['Laba Bersih', year] / df.loc['Total Ekuitas', year] * 100, 2),
        'Current Ratio': round(df.loc['Total Aset Lancar', year] / df.loc['Total Liabilitas Jangka Pendek', year], 2),
        'Debt to Equity (x)': round(df.loc['Total Liabilitas', year] / df.loc['Total Ekuitas', year], 2),
        'Revenue Growth (%)': 'N/A' if year == '2022' else round((df.loc['Pendapatan Bersih', '2023'] / df.loc['Pendapatan Bersih', '2022'] - 1) * 100, 2)
    }

df_rasio = pd.DataFrame(rasio)
print("=== RASIO KEUANGAN MTDL 2022-2023 ===")
print(df_rasio.to_string())
````

````python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('MTDL Financial Ratio Analysis 2022-2023', fontsize=16, fontweight='bold')

years = ['2022', '2023']
x = np.arange(len(years))
width = 0.35

# 1. Profit Margins
axes[0,0].bar(x - width/2, [8.24, 8.65], width, label='Gross PM', color='#2196F3')
axes[0,0].bar(x + width/2, [4.13, 4.26], width, label='Net PM', color='#4CAF50')
axes[0,0].set_title('Profit Margins (%)')
axes[0,0].set_xticks(x)
axes[0,0].set_xticklabels(years)
axes[0,0].legend()
axes[0,0].set_ylabel('%')

# 2. ROA & ROE
axes[0,1].bar(x - width/2, [10.1, 9.27], width, label='ROA', color='#FF9800')
axes[0,1].bar(x + width/2, [19.62, 18.66], width, label='ROE', color='#F44336')
axes[0,1].set_title('ROA & ROE (%)')
axes[0,1].set_xticks(x)
axes[0,1].set_xticklabels(years)
axes[0,1].legend()
axes[0,1].set_ylabel('%')

# 3. Current Ratio & DER
axes[1,0].bar(x - width/2, [1.96, 1.92], width, label='Current Ratio', color='#9C27B0')
axes[1,0].bar(x + width/2, [0.94, 1.01], width, label='DER', color='#FF5722')
axes[1,0].set_title('Likuiditas & Leverage')
axes[1,0].set_xticks(x)
axes[1,0].set_xticklabels(years)
axes[1,0].legend()
axes[1,0].axhline(y=1, color='red', linestyle='--', alpha=0.5, label='Batas DER=1')

# 4. Revenue & Net Profit (Triliun)
rev = [20988297/1000000, 22086390/1000000]
profit = [866721/1000000, 940104/1000000]
axes[1,1].bar(x - width/2, rev, width, label='Revenue (T)', color='#2196F3')
axes[1,1].bar(x + width/2, profit, width, label='Net Profit (T)', color='#4CAF50')
axes[1,1].set_title('Revenue vs Net Profit (Triliun Rp)')
axes[1,1].set_xticks(x)
axes[1,1].set_xticklabels(years)
axes[1,1].legend()
axes[1,1].set_ylabel('Triliun Rp')

plt.tight_layout()
plt.savefig(r'C:\Users\Yoel\Lab_AI\Portofolio\MTDL_ratio_chart.png', dpi=150)
plt.show()
print("Chart tersimpan!")
````
